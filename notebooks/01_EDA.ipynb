```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data_folder = "../data/raw/MachineLearningCVE/"

rows = []

for file in sorted(os.listdir(data_folder)):
    if file.endswith(".csv"):

        df = pd.read_csv(os.path.join(data_folder, file))

        label_col = " Label" if " Label" in df.columns else "Label"

        counts = df[label_col].value_counts()

        for label, count in counts.items():
            rows.append([
                file.replace(".pcap_ISCX.csv",""),
                label,
                count
            ])

plot_df = pd.DataFrame(rows, columns=["Dataset","Classe","Nombre"])

plt.figure(figsize=(18,8))

sns.barplot(
    data=plot_df,
    x="Dataset",
    y="Nombre",
    hue="Classe"
)

plt.xticks(rotation=45, ha="right")
plt.xlabel("Fichiers")
plt.ylabel("Nombre d'instances")
plt.title("Répartition des classes dans les différents fichiers du jeu de données CICIDS2017")

plt.tight_layout()
plt.show()
```
