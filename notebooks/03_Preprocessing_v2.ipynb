```python
import numpy as np

def validate_preprocessed_dataset(df):
    print("=" * 50)
    print("Dataset Validation")
    print("=" * 50)

    print(f"Shape: {df.shape}")

    assert "Label" in df.columns, "❌ Label column missing."
    assert "Label_encoded" not in df.columns, "❌ Label_encoded found."

    missing = df.isnull().sum().sum()
    print(f"Missing values: {missing}")
    assert missing == 0

    numeric_df = df.select_dtypes(include=[np.number])
    infinite = np.isinf(numeric_df).sum().sum()
    print(f"Infinite values: {infinite}")
    assert infinite == 0

    duplicates = df.duplicated().sum()
    print(f"Duplicate rows: {duplicates}")
    assert duplicates == 0

    constant_cols = [
        col for col in df.columns
        if df[col].nunique() == 1
    ]

    print(f"Constant columns: {len(constant_cols)}")
    if constant_cols:
        print(constant_cols)

    print("\nClass distribution:")
    print(df["Label"].value_counts())

    print("\n✅ Dataset validation passed.")
```

```python
validate_preprocessed_dataset(df)
```
